Cuando trabajamos con grandes volúmenes de datos, una de las tareas más comunes es resumirlos (calcular el total, el promedio, el máximo, etc.). Vamos a comparar cómo se hace esto en Python nativo versus NumPy.

La solución nativa: Para sumar una lista en Python sin usar NumPy, estás obligado a usar una función externa llamada sum(). Pruébalo así:

La solución eficiente con NumPy: Transforma la lista en un array y utiliza el método propio de NumPy .sum(). Este método está optimizado para ejecutarse a nivel de memoria de forma ultra rápida:

Python

print(arreglo.sum())

**Ejercicio 2**:
*Para este ejercicio vamos a usar una herramienta de Jupyter llamada %timeit. Este es un comando que analiza la línea de código que tiene a la derecha, la ejecuta miles de veces automáticamente y calcula el tiempo promedio que tardó en completarse.

¿Cómo leer los resultados de %timeit?
Te va a mostrar el tiempo usando diferentes unidades médicas de tiempo:

ms (Milisegundos): La milésima parte de un segundo.

µs (Microsegundos): ¡La millonésima parte de un segundo! (Es decir, muchísimo más rápido).

Consigna: Escribe el código correspondiente en dos celdas separadas de tu cuaderno para comparar cuánta ventaja nos da trabajar con NumPy al procesar un millón de datos.

Celda A: El camino tradicional con Listas Nativas
Crea dos listas independientes llamadas lista_A y lista_B, cada una con los números del 0 al 999.999 usando list(range(1000000)).

Crea una tercera lista vacía llamada lista_resultado.

Escribe un ciclo for que recorra el rango de ese millón de elementos. Dentro del bucle, suma el elemento de la lista_A con el de la lista_B en la misma posición usando sus índices, y agrega el resultado a lista_resultado usando .append().

Al principio de la línea del bucle for, coloca el comando %timeit separado por un espacio para medir cuánto tarda Python en hacer todo ese recorrido.

```python
listaA = list(range(1000000))
listaB = list(range(1000000))
listaResultado = []


%timeit for i in range(len(listaA)):
    listaResultado.append(listaA[i] + listaB[i])
```
