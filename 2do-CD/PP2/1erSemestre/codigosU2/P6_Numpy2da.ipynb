# (Listas vs. Arrays)

En Python nativo, usamos listas que pueden guardar cualquier tipo de objeto (texto, números, booleanos). Sin embargo, para la Ciencia de Datos, procesar millones de registros de esta forma es muy lento. NumPy introduce el ndarray (arreglos de N-dimensiones), bloques de datos del mismo tipo que se almacenan juntos en la memoria del computador, permitiendo cálculos  veloces.

Ejemplo
Python

```python
import numpy as np

# Creación de un vector simple
lista_precios = [10, 20, 30]
array_precios = np.array(lista_precios)

print("Lista con comas:", lista_precios)
print("Array NumPy (notar que no tiene comas):", array_precios)
```

# (Vectorización)

Si queremos sumarle un valor a cada elemento de una lista de Python, estamos obligados a escribir un ciclo for que recorra la estructura elemento por elemento. En NumPy existe la vectorización: las operaciones matemáticas se aplican a todo el bloque al mismo tiempo directamente a nivel de hardware.

Podemos anteponer el comando  %timeit al principio de una celda para medir de forma real cuántos milisegundos (ms) o microsegundos (µs) tarda en ejecutarse un bloque de código.

Otra forma de medir de forma exacta cuánto tarda en ejecutarse un bloque de código en cualquier entorno (ya sea una celda de Google Colab o un script en Visual Studio Code), podemos usar la librería nativa de Python llamada time.

La estrategia consiste en guardar la hora actual antes de que empiece el cálculo con time.time(), ejecutar el proceso y volver a tomar la hora al finalizar. Al restar el tiempo final menos el tiempo inicial, obtenemos los segundos exactos que tardó la computadora en procesar la operación.

```python
#Ejemplo
import time

inicio = time.time()
# ... aquí va el proceso que queremos medir ...
fin = time.time()

tiempo_total = fin - inicio
print("El proceso tardó:",tiempo_total, "segundos.")
```

# Estructuración de Datos en Matrices (Shape & Reshape)

Los datos tabulares del mundo real (filas y columnas) se representan en NumPy como arrays bidimensionales o matrices. Todo array posee una propiedad fundamental llamada .shape, que nos devuelve una tupla con sus dimensiones (filas, columnas). Podemos cambiar la forma o estructura de un array mediante el método .reshape() sin alterar los datos internos.

Ejemplo

```python
import numpy as np
# Creamos un vector de 6 elementos y lo reestructuramos en una matriz de 2x3
vector = np.arange(6)
matriz = vector.reshape(2, 3)
print(matriz)
print("Dimensiones:", matriz.shape)
```

# Segmentación e Indexación Condicional (Máscaras Booleanas)

El PDF destaca una de las herramientas más potentes de NumPy: la capacidad de filtrar elementos utilizando condiciones lógicas (operadores como >, <, ==). Al aplicar una condición a un array, NumPy evalúa cada celda y genera una "máscara booleana" (de valores True o False). Si pasamos esa máscara al array original, se extraerán únicamente los elementos que cumplan la condición.

Ejemplo

```python
import numpy as np
edades = np.array([20, 45, 18, 30])
# edades = [20, 45, 18, 30]
mayores_de_25 = edades[edades > 25]
print(mayores_de_25)  # Devuelve: [45, 30]
```

# (Crecimiento Secuencial de Fibonacci)

En la naturaleza y la agricultura, muchas estructuras siguen patrones secuenciales de crecimiento como la famosa Sucesión de Fibonacci. En este paso final, analizaremos cómo se comporta la relación entre términos contiguos para observar cómo converge hacia el Número Áureo ($\Phi \approx 1.618$).

Ejemplo

Si tenemos una lista de Fibonacci inicial [0, 1, 1, 2, 3, 5, 8], podemos acceder a los últimos elementos usando índices negativos (por ejemplo, fibo[-1] para el último valor).

# (Ejercicio 9 - Desafío Final)

Asumiendo que ya posees una lista con los primeros 30 números de la sucesión de Fibonacci llamada fibo:Crea un bucle que recorra los últimos 5 pares de números contiguos de tu lista de datos.En cada iteración, calcula el cociente dividiendo el término actual por su término inmediatamente anterior ($n_i / n_{i-1}$) e imprime los resultados en pantalla.

Pregunta de cierre: Observando los 5 resultados impresos en la consola, ¿a qué valor decimal fijo se aproximan de forma matemática constante?

```python
fibo = [0, 1]

for _ in range(28):
    nextNumber = fibo[-2] + fibo[-1]
    fibo.append(nextNumber)  

# Creamos el bucle para recorrer los índices desde el 25 hasta el 29
for i in range(25, 30):
    termino_actual = fibo[i]
    termino_anterior = fibo[i - 1]
    
    # Calculamos el cociente
    cociente = termino_actual / termino_anterior
    
    # Imprimimos los resultados en pantalla
    print(f"División de {termino_actual} / {termino_anterior} = {cociente}")
```
