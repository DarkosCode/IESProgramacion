## Paso 12 — Establecer un criterio para eliminar filas

Eliminar cualquier fila que contenga un dato faltante puede provocar una pérdida importante de información.

Una alternativa consiste en establecer un criterio.

El parámetro `thresh` permite indicar una cantidad mínima de valores que deben estar presentes.

Por ejemplo:

```python
copia.dropna(thresh=3)
```

conserva las filas que tienen al menos tres valores no faltantes.

**Consigna**

Crear una copia de `wine_reviews` y establecer un criterio que permita conservar únicamente las filas que tengan **al menos tres valores completos**.

Determinar cuántas instancias quedan después de aplicar el criterio.

````python
# Copia del dataset original
copia_thresh = wine_reviews.copy()

# Conservar únicamente las filas que tengan al menos 3 valores completos
copia_thresh = copia_thresh.dropna(thresh=3)

# Cantidad de instancias que quedan
print("Instancias luego de aplicar el criterio thresh=3:", copia_thresh.shape[0])
````

## Paso 13 — Eliminar columnas con demasiados valores faltantes

También puede establecerse un criterio para decidir cuándo una columna tiene demasiados valores faltantes.

El atributo `shape[0]` permite conocer la cantidad de filas de un DataFrame:

```python
datos.shape[0]
```

Por lo tanto, el 50 % de las filas puede expresarse como:

```python
datos.shape[0] * 0.5
```

El parámetro `thresh` de `dropna()` permite utilizar este tipo de criterio.

**Consigna**

1. Crear una copia de `wine_reviews`.
2. Eliminar las columnas que tengan más del 50 % de sus valores faltantes.
3. Determinar cuántas columnas quedan.
4. Identificar cuáles son los atributos que permanecen.
5. Luego, eliminar las instancias que todavía tengan algún valor faltante.
6. Determinar cuántas instancias quedan finalmente.

````python
# Copia del dataset original
copia_cols_50 = wine_reviews.copy()

# Cantidad mínima de valores no faltantes para conservar una columna (50% de las filas)
minimo = copia_cols_50.shape[0] * 0.5

# Eliminar columnas con más del 50% de valores faltantes
copia_cols_50 = copia_cols_50.dropna(axis=1, thresh=minimo)

print("Columnas que quedan:", copia_cols_50.shape[1])
print("Atributos que permanecen:", list(copia_cols_50.columns))

# Eliminar las instancias que todavía tengan algún valor faltante
copia_cols_50 = copia_cols_50.dropna()

print("Instancias finales:", copia_cols_50.shape[0])
````

# Imputación de valores faltantes

Eliminar datos no siempre es la mejor alternativa.

Otra posibilidad consiste en **imputar** los valores faltantes, es decir, reemplazarlos utilizando algún valor estimado.

En este caso se trabajará especialmente con la columna `price`.

Antes de imputar valores es necesario conocer cómo se distribuyen los datos disponibles.

## Paso 15 — Calcular medidas estadísticas

Para describir una variable numérica pueden utilizarse diferentes medidas.

En Pandas:

**Media**

```python
datos["price"].mean()
```

**Mediana**

```python
datos["price"].median()
```

**Moda**

```python
datos["price"].mode()
```

**Desviación estándar**

```python
datos["price"].std()
```

La media representa el promedio de los valores. La mediana corresponde al valor central de los datos ordenados. La moda representa el valor que aparece con mayor frecuencia. La desviación estándar permite describir la dispersión de los datos respecto de la media.

**Consigna**

Calcular para la columna `price`:

1. La media.
2. La moda.
3. La mediana.
4. La desviación estándar.

Registrar los resultados para utilizarlos en los pasos siguientes.

````python
media_price = wine_reviews["price"].mean()
moda_price = wine_reviews["price"].mode()
mediana_price = wine_reviews["price"].median()
std_price = wine_reviews["price"].std()

print("Media:", media_price)
print("Moda:", moda_price.tolist())
print("Mediana:", mediana_price)
print("Desviación estándar:", std_price)
````

## Paso 16 — Imputar utilizando la media

El método `fillna()` permite reemplazar los valores faltantes.

Por ejemplo:

```python
datos["columna"].fillna(valor)
```

También puede utilizarse sobre un DataFrame completo:


## Paso 17 — Analizar el efecto de la imputación

Imputar valores modifica los datos y, por lo tanto, puede modificar las medidas estadísticas.

**Consigna**

Volver a calcular sobre el dataset imputado:

- media;
- mediana;
- moda;
- desviación estándar.

Comparar estos resultados con los obtenidos antes de la imputación.

Responder:

1. ¿Cambió la media?
2. ¿Cambió la mediana?
3. ¿Cambió la moda?
4. ¿Cambió la desviación estándar?
5. ¿Por qué algunos estadísticos pueden cambiar como consecuencia de la imputación?

````python
media_imputada = wine_reviews_media["price"].mean()
mediana_imputada = wine_reviews_media["price"].median()
moda_imputada = wine_reviews_media["price"].mode()
std_imputada = wine_reviews_media["price"].std()

print("Media original:", media_price, "| Media imputada:", media_imputada)
print("Mediana original:", mediana_price, "| Mediana imputada:", mediana_imputada)
print("Moda original:", moda_price.tolist(), "| Moda imputada:", moda_imputada.tolist())
print("Desv. estándar original:", std_price, "| Desv. estándar imputada:", std_imputada)

# Respuestas:
# 1. La media no cambia, porque se imputó justamente utilizando la propia media:
#    al agregar múltiples valores iguales al promedio, el promedio se mantiene igual.
# 2. La mediana puede cambiar, ya que se agregan muchos valores nuevos e idénticos
#    (la media) que pueden desplazar el valor central del conjunto de datos.
# 3. La moda puede cambiar (o incluso puede pasar a ser la propia media),
#    porque ahora ese valor aparece una gran cantidad de veces adicionales.
# 4. La desviación estándar suele disminuir, ya que se agregan muchos valores
#    idénticos y cercanos al promedio, lo que reduce la dispersión general de los datos.
# 5. Estos cambios ocurren porque la imputación no es un dato real observado,
#    sino un valor "artificial" repetido muchas veces, lo que altera la forma
#    de la distribución y, por lo tanto, las medidas estadísticas que dependen
#    de ella (especialmente la mediana, la moda y la desviación estándar).
````

## Paso 18 — Comparar diferentes estrategias de imputación

La media no es la única alternativa posible para reemplazar valores faltantes.

También pueden utilizarse la mediana o la moda.

**Consigna**

Realizar nuevamente el procedimiento de imputación, utilizando:

1. La **mediana** de `price`.
2. La **moda** de `price`.

Para cada estrategia:

- crear una copia independiente del dataset;
- realizar la imputación;
- calcular nuevamente media, mediana, moda y desviación estándar;
- comparar los resultados con los obtenidos mediante la imputación por la media.

Finalmente, indicar qué diferencias observa entre las tres estrategias.

````python
# Imputación utilizando la mediana
wine_reviews_mediana = wine_reviews.copy()
wine_reviews_mediana["price"] = wine_reviews_mediana["price"].fillna(mediana_price)

media_mediana = wine_reviews_mediana["price"].mean()
mediana_mediana = wine_reviews_mediana["price"].median()
moda_mediana = wine_reviews_mediana["price"].mode()
std_mediana = wine_reviews_mediana["price"].std()

# Imputación utilizando la moda
wine_reviews_moda = wine_reviews.copy()
valor_moda = moda_price.iloc[0]
wine_reviews_moda["price"] = wine_reviews_moda["price"].fillna(valor_moda)

media_moda = wine_reviews_moda["price"].mean()
mediana_moda = wine_reviews_moda["price"].median()
moda_moda = wine_reviews_moda["price"].mode()
std_moda = wine_reviews_moda["price"].std()

print("=== Imputación por media ===")
print("Media:", media_imputada, "| Mediana:", mediana_imputada, "| Moda:", moda_imputada.tolist(), "| Std:", std_imputada)

print("\n=== Imputación por mediana ===")
print("Media:", media_mediana, "| Mediana:", mediana_mediana, "| Moda:", moda_mediana.tolist(), "| Std:", std_mediana)

print("\n=== Imputación por moda ===")
print("Media:", media_moda, "| Mediana:", mediana_moda, "| Moda:", moda_moda.tolist(), "| Std:", std_moda)

# Diferencias observadas:
# - La imputación por la media mantiene la media original, pero puede distorsionar
#   la mediana y sobre todo reducir la desviación estándar.
# - La imputación por la mediana suele modificar menos la mediana en sí (por definición
#   se refuerza el valor central), pero cambia la media si la mediana difiere de ella,
#   y también tiende a reducir la desviación estándar.
# - La imputación por la moda es la que más distorsiona la distribución cuando la moda
#   está lejos de la media/mediana, ya que agrega muchas repeticiones de un único valor,
#   lo que puede modificar fuertemente la moda resultante y reducir aún más la varianza.
# - En los tres casos, la desviación estándar disminuye respecto de la original, porque
#   se están agregando valores "artificiales" que no representan la variabilidad real
#   de los datos faltantes.
````

````python
# Actividad integradora: análisis completo del dataset wine_reviews

# 1. Carga y exploración inicial del dataset
wine_reviews_final = pd.read_csv("wine_reviews.csv")
wine_reviews_final = wine_reviews_final.drop(columns=["Unnamed: 0"], errors="ignore")
wine_reviews_final.head()

# 2. Identificación de filas y columnas
print("Filas:", wine_reviews_final.shape[0])
print("Columnas:", wine_reviews_final.shape[1])
print("Nombres de columnas:", list(wine_reviews_final.columns))

# 3. Identificación de valores faltantes
print("\nValores faltantes por columna:")
print(wine_reviews_final.isna().sum())

# 4. Estrategia de tratamiento de valores faltantes:
# se imputa "price" con la mediana, ya que es una medida más robusta frente a
# valores extremos que la media, y se eliminan las filas que aún tengan
# valores faltantes en columnas clave como "variety" o "country".
wine_reviews_final["price"] = wine_reviews_final["price"].fillna(
    wine_reviews_final["price"].median()
)
wine_reviews_final = wine_reviews_final.dropna(subset=["variety", "country"])

# 5. Medidas estadísticas para una variable numérica (price)
print("\nEstadísticas de price luego del tratamiento:")
print("Media:", wine_reviews_final["price"].mean())
print("Mediana:", wine_reviews_final["price"].median())
print("Moda:", wine_reviews_final["price"].mode().tolist())
print("Desviación estándar:", wine_reviews_final["price"].std())

# 6. Visualización
plt.hist(wine_reviews_final["price"], bins=40)
plt.xscale("log")
plt.xlabel("Precio (price) - escala logarítmica")
plt.ylabel("Frecuencia")
plt.title("Distribución de precios luego de la imputación")
plt.grid()
plt.show()

# 7. Conclusión:
# El dataset original presentaba una cantidad considerable de valores faltantes,
# concentrados principalmente en columnas como "region_2", "designation" y
# "taster_name". Dado que "price" es una variable numérica relevante, se optó
# por imputar sus valores faltantes con la mediana en lugar de eliminarlos,
# evitando así una pérdida importante de instancias. Para columnas categóricas
# clave como "variety" y "country", que presentan muy pocos valores faltantes,
# se optó por eliminar directamente esas filas, ya que la pérdida de información
# es mínima. El histograma en escala logarítmica confirma que los precios están
# fuertemente concentrados en valores bajos, con una cola de vinos de precio muy
# elevado, lo que justifica el uso de esta escala para su correcta interpretación.
````
