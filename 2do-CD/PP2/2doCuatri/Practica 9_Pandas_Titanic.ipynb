# Trabajando con valores faltantes

En un dataset real puede ocurrir que determinada información no haya sido registrada.

Pandas representa habitualmente los valores faltantes mediante `NaN`.

El método:

```python
datos.isna()
```

permite identificar las celdas que contienen valores faltantes.

El resultado utiliza `True` cuando el dato está ausente y `False` cuando está presente.

## Paso 6 — Contar valores faltantes

Para conocer cuántos valores faltantes hay en cada columna podemos combinar `isna()` con `sum()`:

```python
datos.isna().sum()
```

**Consigna**

## Paso 7 — Identificar la columna con más valores faltantes

Ya conocemos cómo contar los valores faltantes por columna.

El método `idxmax()` permite obtener el índice o nombre asociado al valor máximo.

Por ejemplo:

```python
datos.isna().sum().idxmax()
```

## Paso 8 — Analizar los valores faltantes por fila

Hasta ahora se analizaron los valores faltantes por columna. También podemos contar cuántos valores faltantes tiene cada pasajero.

En Pandas, `axis=1` indica que la operación se realiza considerando las columnas de cada fila.

Por ejemplo:

Para localizar la fila con la mayor cantidad de valores faltantes se puede utilizar:

# Imputación de valores faltantes

Eliminar datos no siempre es la mejor alternativa.

Antes de elegir un valor para reemplazar los datos faltantes, es conveniente conocer algunas medidas estadísticas.

## Paso 14 — Calcular medidas estadísticas de `Age`

Para una variable numérica podemos calcular diferentes medidas.

**Media**

```python
datos["Age"].mean()
```

**Moda**

```python
datos["Age"].mode()
```

**Mediana**

```python
datos["Age"].median()
```

**Desviación estándar**

```python
datos["Age"].std()
```

La media representa el promedio. La mediana corresponde al valor central de los datos ordenados. La moda representa el valor que aparece con mayor frecuencia. La desviación estándar permite analizar la dispersión de los valores respecto de la media.

# Análisis de supervivencia

Ahora se utilizarán algunas herramientas de Pandas para investigar si la supervivencia estuvo relacionada con diferentes características de los pasajeros.

La variable `Survived` contiene valores 0 y 1. Por lo tanto, su **media puede interpretarse como una proporción de supervivientes**.

Por ejemplo, si para un grupo la media de `Survived` es `0.60`, significa que aproximadamente el 60 % de ese grupo sobrevivió.

## Paso 20 — Agrupar por clase

El método `groupby()` permite dividir los datos en grupos según una variable.

Por ejemplo:

```python
datos.groupby("Pclass")
```

Podemos seleccionar las columnas que nos interesan y calcular la media:

```python
datos[["Pclass", "Survived"]].groupby("Pclass").mean()
```

Finalmente, `sort_values()` permite ordenar los resultados:

```python
datos[["Pclass", "Survived"]].groupby("Pclass").mean().sort_values(
    by="Survived", ascending=False
)
```
