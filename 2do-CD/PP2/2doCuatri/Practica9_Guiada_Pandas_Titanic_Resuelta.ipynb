## Paso 12 — Establecer un criterio para eliminar filas

Eliminar cualquier fila que contenga un valor faltante puede provocar una pérdida importante de información.

Una alternativa es establecer un criterio mínimo de datos completos.

El parámetro `thresh` permite indicar la cantidad mínima de valores **no faltantes** que debe conservar una fila.

Por ejemplo:

```python
copia.dropna(thresh=5)
```

conserva las filas que tienen al menos cinco valores completos.

**Consigna**

El objetivo es descartar aquellas filas que tengan **más de tres valores faltantes**.

Para lograrlo, determine primero cuántas columnas tiene el dataset y, a partir de ese dato, establezca el valor apropiado para `thresh`.

Luego:

1. Cree una copia de `titanic`.
2. Aplique el criterio.
3. Determine cuántas instancias quedan.

````python
# Cantidad total de columnas
total_columnas = titanic.shape[1]

# Para descartar filas con más de 3 valores faltantes, se deben conservar
# aquellas filas que tengan al menos (total_columnas - 3) valores completos
thresh = total_columnas - 3

# 1. Copia del dataset
copia_thresh = titanic.copy()

# 2. Aplicar el criterio
copia_thresh = copia_thresh.dropna(thresh=thresh)

# 3. Cantidad de instancias que quedan
print("Thresh utilizado:", thresh)
print("Instancias que quedan:", copia_thresh.shape[0])
````

El parámetro `thresh` puede utilizarse junto con `axis=1` para establecer un mínimo de valores completos por columna.

**Consigna**

# Imputación de valores faltantes

Eliminar datos no siempre es la mejor alternativa.

Otra estrategia consiste en **imputar** los valores faltantes: reemplazarlos por un valor elegido según algún criterio.

En esta práctica se trabajará principalmente con la columna `Age`.

Antes de elegir un valor para reemplazar los datos faltantes, es conveniente conocer algunas medidas estadísticas.

## Paso 14 — Calcular medidas estadísticas de `Age`

Para una variable numérica podemos calcular diferentes medidas.

**Media**

```python
datos["Age"].mean()
```

**Moda**

```python
datos["Age"].mode()
```

**Mediana**

```python
datos["Age"].median()
```

**Desviación estándar**

```python
datos["Age"].std()
```

La media representa el promedio. La mediana corresponde al valor central de los datos ordenados. La moda representa el valor que aparece con mayor frecuencia. La desviación estándar permite analizar la dispersión de los valores respecto de la media.

**Consigna**

Calcular para la columna `Age`:

1. La media.
2. La moda.
3. La mediana.
4. La desviación estándar.

Registrar los resultados para utilizarlos posteriormente.

````python
media_age = titanic["Age"].mean()
moda_age = titanic["Age"].mode()
mediana_age = titanic["Age"].median()
std_age = titanic["Age"].std()

print("Media:", media_age)
print("Moda:", moda_age.tolist())
print("Mediana:", mediana_age)
print("Desviación estándar:", std_age)
````

## Paso 15 — Imputar valores utilizando la media

El método `fillna()` permite reemplazar valores faltantes.

Por ejemplo:

```python
datos["Age"].fillna(valor)
```

También puede modificarse el DataFrame directamente mediante:

```python

```

**Consigna**

1. Crear una copia de `titanic`.

````python
# 1. Copia del dataset
titanic_media = titanic.copy()

# 2. Reemplazar los valores faltantes de Age utilizando la media
titanic_media["Age"] = titanic_media["Age"].fillna(media_age)

# 3. Verificar que ya no queden valores faltantes en Age
print("Valores faltantes en Age (copia imputada):", titanic_media["Age"].isna().sum())

# 4. El dataset original no fue modificado
print("Valores faltantes en Age (original):", titanic["Age"].isna().sum())
````

## Paso 16 — Analizar el efecto de la imputación

La imputación modifica algunos valores del dataset y puede modificar las medidas estadísticas.

**Consigna**

Sobre el dataset después de imputar `Age` con la media, volver a calcular:

- media;
- moda;
- mediana;
- desviación estándar.

Comparar los resultados con los obtenidos antes de la imputación.

Responder:

1. ¿Cambió la media?
2. ¿Cambió la mediana?
3. ¿Cambió la moda?
4. ¿Cambió la desviación estándar?
5. ¿Por qué se produjeron esos cambios?

````python
media_imputada = titanic_media["Age"].mean()
mediana_imputada = titanic_media["Age"].median()
moda_imputada = titanic_media["Age"].mode()
std_imputada = titanic_media["Age"].std()

print("Media original:", media_age, "| Media imputada:", media_imputada)
print("Mediana original:", mediana_age, "| Mediana imputada:", mediana_imputada)
print("Moda original:", moda_age.tolist(), "| Moda imputada:", moda_imputada.tolist())
print("Desv. estándar original:", std_age, "| Desv. estándar imputada:", std_imputada)

# Respuestas:
# 1. La media no cambia, ya que se imputó justamente utilizando la propia media:
#    agregar múltiples valores iguales al promedio no modifica el promedio.
# 2. La mediana puede cambiar, porque se agregan muchos valores nuevos e idénticos
#    (la media) que pueden desplazar el valor central del conjunto de datos.
# 3. La moda puede cambiar (e incluso puede pasar a ser la propia media), ya que
#    ese valor ahora aparece una cantidad adicional de veces.
# 4. La desviación estándar suele disminuir, porque se agregan muchos valores
#    idénticos y cercanos al promedio, reduciendo la dispersión general.
# 5. Estos cambios ocurren porque la imputación agrega un valor "artificial"
#    repetido muchas veces, lo que altera la forma de la distribución original
#    y, por lo tanto, las medidas estadísticas que dependen de ella.
````

## Paso 17 — Comparar media, mediana y moda como estrategias

La media no es la única posibilidad para reemplazar valores faltantes.

También podemos utilizar la mediana:

```python
datos["Age"].median()
```

o la moda:

```python
datos["Age"].mode()[0]
```

**Consigna**

Crear dos nuevas copias independientes de `titanic`.

En la primera:

- imputar los valores faltantes de `Age` utilizando la **mediana**.

En la segunda:

- imputar los valores faltantes de `Age` utilizando la **moda**.

Para cada caso, volver a calcular los cuatro estadísticos trabajados.

Comparar las tres estrategias: media, mediana y moda.

Indicar qué diferencias observa.

````python
# Imputación utilizando la mediana
titanic_mediana = titanic.copy()
titanic_mediana["Age"] = titanic_mediana["Age"].fillna(mediana_age)

media_mediana = titanic_mediana["Age"].mean()
mediana_mediana = titanic_mediana["Age"].median()
moda_mediana = titanic_mediana["Age"].mode()
std_mediana = titanic_mediana["Age"].std()

# Imputación utilizando la moda
titanic_moda = titanic.copy()
valor_moda = moda_age[0]
titanic_moda["Age"] = titanic_moda["Age"].fillna(valor_moda)

media_moda = titanic_moda["Age"].mean()
mediana_moda = titanic_moda["Age"].median()
moda_moda = titanic_moda["Age"].mode()
std_moda = titanic_moda["Age"].std()

print("=== Imputación por media ===")
print("Media:", media_imputada, "| Mediana:", mediana_imputada, "| Moda:", moda_imputada.tolist(), "| Std:", std_imputada)

print("\n=== Imputación por mediana ===")
print("Media:", media_mediana, "| Mediana:", mediana_mediana, "| Moda:", moda_mediana.tolist(), "| Std:", std_mediana)

print("\n=== Imputación por moda ===")
print("Media:", media_moda, "| Mediana:", mediana_moda, "| Moda:", moda_moda.tolist(), "| Std:", std_moda)

# Diferencias observadas:
# - La imputación por la media mantiene la media original, pero puede distorsionar
#   la mediana y reducir la desviación estándar.
# - La imputación por la mediana refuerza el valor central, pero puede modificar
#   la media si esta difiere de la mediana, y también reduce la dispersión.
# - La imputación por la moda es la que más distorsiona la distribución cuando la
#   moda está lejos de la media/mediana, ya que agrega muchas repeticiones de un
#   único valor.
# - En los tres casos, la desviación estándar disminuye respecto de la original,
#   porque se agregan valores "artificiales" que no representan la variabilidad
#   real de los datos faltantes.
````

# Análisis de supervivencia

Ahora se utilizarán algunas herramientas de Pandas para investigar si la supervivencia estuvo relacionada con diferentes características de los pasajeros.

La variable `Survived` contiene valores 0 y 1. Por lo tanto, su **media puede interpretarse como una proporción de supervivientes**.

Por ejemplo, si para un grupo la media de `Survived` es `0.60`, significa que aproximadamente el 60 % de ese grupo sobrevivió.

## Paso 20 — Agrupar por clase

El método `groupby()` permite dividir los datos en grupos según una variable.

Por ejemplo:

```python
datos.groupby("Pclass")
```

Podemos seleccionar las columnas que nos interesan y calcular la media:


A partir de los análisis realizados, responder:

> **¿La supervivencia de los pasajeros del Titanic parece haber sido completamente aleatoria o existen características que muestran una relación con la supervivencia?**

La respuesta debe estar fundamentada en los resultados obtenidos mediante Pandas.

**Importante:** encontrar una relación entre dos variables no significa necesariamente que una sea la causa de la otra. Diferenciar entre una asociación observada en los datos y una explicación causal.

````python
# Actividad integradora: análisis exploratorio completo del dataset Titanic

# 1. Exploración de la estructura del dataset
titanic_final = pd.read_csv("Titanic.csv")
print("Filas:", titanic_final.shape[0])
print("Columnas:", titanic_final.shape[1])
print("Nombres de columnas:", list(titanic_final.columns))
titanic_final.info()

# 2. Identificación de valores faltantes
print("\nValores faltantes por columna:")
print(titanic_final.isna().sum())

# 3. Estrategia justificada para tratar los valores faltantes:
# - "Cabin" tiene una proporción muy alta de valores faltantes y se elimina.
# - "Age" es una variable numérica relevante para el análisis y se imputa
#   utilizando la mediana, que es más robusta frente a valores extremos.
# - "Embarked" tiene muy pocos valores faltantes, por lo que se eliminan
#   directamente esas filas.
titanic_final = titanic_final.drop(columns=["Cabin"], errors="ignore")
titanic_final["Age"] = titanic_final["Age"].fillna(titanic_final["Age"].median())
titanic_final = titanic_final.dropna(subset=["Embarked"])

# 4. Medidas estadísticas para Age
print("\nEstadísticas de Age luego del tratamiento:")
print("Media:", titanic_final["Age"].mean())
print("Mediana:", titanic_final["Age"].median())
print("Moda:", titanic_final["Age"].mode().tolist())
print("Desviación estándar:", titanic_final["Age"].std())

# 5. Histograma
plt.hist(titanic_final["Age"], bins=10)
plt.xlabel("Edad")
plt.ylabel("Frecuencia")
plt.title("Distribución de edades luego del tratamiento de faltantes")
plt.grid()
plt.show()

# 6. Creación de Rango_Edad
titanic_final["Rango_Edad"] = pd.cut(titanic_final["Age"], 5)

# 7. Análisis de supervivencia utilizando groupby()
print("\nSupervivencia según clase:")
print(titanic_final[["Pclass", "Survived"]].groupby("Pclass").mean().sort_values(
    by="Survived", ascending=False
))

print("\nSupervivencia según sexo:")
print(titanic_final[["Sex", "Survived"]].groupby("Sex").mean().sort_values(
    by="Survived", ascending=False
))

# 8. Análisis combinando dos o más variables
print("\nSupervivencia según sexo y clase:")
print(titanic_final[["Sex", "Pclass", "Survived"]].groupby(["Sex", "Pclass"]).mean().sort_values(
    by="Survived", ascending=False
))

# Pregunta final:
# La supervivencia de los pasajeros del Titanic no parece haber sido aleatoria.
# Los análisis realizados muestran diferencias claras y consistentes: las
# mujeres presentaron una proporción de supervivencia mucho mayor que los
# varones, los pasajeros de primera clase sobrevivieron en mayor proporción
# que los de tercera clase, y los pasajeros más jóvenes tendieron a sobrevivir
# más que los de mayor edad. Estas diferencias se mantienen e incluso se
# potencian al combinar variables (por ejemplo, mujeres de primera clase).
# Sin embargo, que existan estas asociaciones no implica necesariamente una
# relación causal directa: variables como el sexo o la clase pudieron estar
# vinculadas a otros factores (por ejemplo, la ubicación del camarote o el
# orden de acceso a los botes salvavidas) que explican en mayor medida la
# supervivencia observada.
````
