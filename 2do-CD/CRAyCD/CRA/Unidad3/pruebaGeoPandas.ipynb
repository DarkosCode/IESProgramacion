cuando creás un mapa con folium.Map(), la librería se conecta a los servidores de OpenStreetMap (un proyecto mundial de mapeo colaborativo y gratuito, similar a Wikipedia pero de mapas)

```python
import geopandas as gpd
import folium

# 1. Cargamos el GeoJSON oficial de IDE Mendoza
mapa_godoy_cruz = gpd.read_file("DepartamentosResultado.geojson")

# 2. Creamos el mapa base con un estilo analítico gris claro ("CartoDB positron")
m = folium.Map(location=[-32.9300, -68.8600], zoom_start=12, tiles="CartoDB positron")

# 3. Dibujamos las líneas del mapa sobre el fondo interactivo
folium.GeoJson(
    mapa_godoy_cruz,
    name="Límites de Godoy Cruz",  # Este nombre aparecerá en el control de capas
    style_function=lambda x: {
        'fillColor': '#2471A3',  
        'color': '#1A5276',      
        'weight': 2,
        'fillOpacity': 0.4
    },
    # Tooltip: se muestra al pasar el mouse
    tooltip=folium.GeoJsonTooltip(
        fields=["departamen"], 
        aliases=["Departamento:"]
    ),
    # NUEVO -> Popup: se muestra al hacer CLIC e incluye más datos de la tabla
    popup=folium.GeoJsonPopup(
        fields=["departamen", "codigo_dep", "fid"],
        aliases=["Departamento:", "Código ID:", "Feature ID:"]
    )
).add_to(m)

# NUEVO -> Control de capas: agrega un botón interactivo para apagar/prender los límites
folium.LayerControl().add_to(m)

# 4. Guardamos el resultado refinado
m.save("mapa_godoy_cruz.html")
```
